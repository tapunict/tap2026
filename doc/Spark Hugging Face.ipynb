# Spark HuggingFace
HuggingFace is a popular open-source library of pre-trained models for artificial intelligence (AI) and natural language processing (NLP) tasks.

- Provides access to a wide range of transformer models
- Supports various NLP tasks such as text classification, named entity recognition, question answering, and more

## Spark HuggingFace Docker

:::: {.columns}
::: {.column width="50%"}
Let's create custom Docker image for Spark HuggingFace

```docker build -t tap:sparkhg .```

```Dockerfile
FROM apache/spark

USER root

# Setup Hugging Face libraries
RUN pip install --no-cache-dir \
    pandas \
    pyarrow \
    transformers \
    torch --extra-index-url https://download.pytorch.org/whl/cpu

# Create cache directories for Hugging Face models
RUN mkdir -p /opt/spark/.cache/huggingface && \
    chown -R spark:spark /opt/spark/.cache

# Set environment variable for Hugging Face cache
ENV HF_HOME=/opt/spark/.cache/huggingface

USER spark
```
:::

::: {.column width="50%"}

To run the image with persistent cache for models and dependencies:
```bash 
docker run   -v ./src:/opt/spark/work-dir   -v huggingface-cache:/opt/spark/.cache/huggingface   tap:sparkhg   /opt/spark/bin/spark-submit sentiment.py
```
