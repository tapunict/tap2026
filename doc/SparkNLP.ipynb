# Spark NLP
SparkNLP: Natural Language Processing (NLP) library for Apache Spark

- Built on top of Apache Spark MLlib
- Provides state-of-the-art NLP algorithms and pre-trained models
- Scalable and efficient for large datasets 

## Spark NLP Docker

:::: {.columns}
::: {.column width="50%"}
Let's create custom Docker image with Spark NLP installed.
```Dockerfile
FROM python:3.10-bullseye

# Install Java 11 
RUN apt-get update && \
    apt-get install -y --no-install-recommends openjdk-11-jdk && \
    apt-get clean && \
    rm -rf /var/lib/apt/lists/*

# Set JAVA_HOME
ENV JAVA_HOME=/usr/lib/jvm/java-11-openjdk-amd64
ENV PATH=$JAVA_HOME/bin:$PATH

# See https://sparknlp.org/docs/en/install#python
# Install Spark NLP and PySpark with dependencies
RUN pip install numpy spark-nlp==6.2.0 pyspark==3.3.1

# Copy application source code
COPY src /usr/src/app

# Set working directory
WORKDIR /usr/src/app

# Set entrypoint to allow passing script name dynamically
ENTRYPOINT ["python3"]
CMD ["example.py"]
```
:::

::: {.column width="50%"}

To run the image with persistent cache for models and dependencies:
```bash 
docker run -p 4040:4040  -e TF_CPP_MIN_LOG_LEVEL=3 -e PYTHONUNBUFFERED=1 -e SPARK_LOG_LEVEL=WARN -v sparknlp-ivy-cache:/root/.ivy2   -v sparknlp-models:/root/cache_pretrained -it -v ./src:/usr/src/app tap:sparknlp python example.py
```

## Sentiment Analysis Example

